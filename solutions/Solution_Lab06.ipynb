## Exercise 1: String replacing

Create a function called `modify_string` that takes a string `text` and two more strings `word_to_find` and `word_to_replace`. The function should:
1. Convert the entire `text` to lowercase.
2. Find the index of `word_to_find` in the `text`.
3. Replace all occurrences of `word_to_find` with `word_to_replace` and return the modified string along with the index.

```python
def modify_string(text, word_to_find, word_to_replace):
    text = text.lower()
    index = text.find(word_to_find)
    modified_text = text.replace(word_to_find, word_to_replace)
    return modified_text, index
```

```python
# Example usage
result, index = modify_string("Hello World! The world is beautiful.", "world", "earth")
print(result)  # Output: "hello earth! the earth is beautiful."
print(index)  # Output: 6
```

## Exercise 2: My final grade

Write a function called `mygrade` that computes the grade that you will get at the end of this Master's degree. It should take the following parameters:
- base: the scale to which you want to convert the grade (usually 110)
- inTime: a boolean that specifies whether you graduated in time or not; consider that this gives you 1 extra point on average scaled to 110
- thesisGrade: a number from 0 to 6 that must be added at the end (this is how much your thesis will be evaluated)
- grades: a sequence (of unknown length) of grades, which must be either numbers between 18 and 30 or a string written like "30L"

Assume that all courses weight the same amount of CFUs. Also, consider that any 30L grants you +0.33 on the average scaled to 110.

The formula would be: `finalGrade = ( average(grades)/30*base ) + (0.33 * number of laudes) + (1 if inTime) + thesisGrade`

```python
def mygrade(base, inTime, thesisGrade, *grades):
    if len(grades) == 0:
        print("At least one grade must be provided.")
        return

    grade_sum = 0
    num_laudes = 0
    for grade in grades:
        if grade == "30L":
            grade_sum += 30
            num_laudes += 1
        elif isinstance(grade, (int, float)) and 18 <= grade <= 30:
            grade_sum += grade
        else:
            print("Grades must be numbers between 18 and 30 or '30L'.")
            return

    average_grade = grade_sum / len(grades)
    final_grade = average_grade / 30 * base + 0.33 * num_laudes
    if inTime:
        final_grade += 1
    return final_grade + thesisGrade # In practice, the final grade cannot exceed the base; if you get something > 100, then you usually get honors (laude)!
```

```python
# Example usage:
print(mygrade(110, True, 4, 30, 28, "30L", 29, 30))
```

## Exercise 3: Check exam admission

A university course requires students to attend a minimum percentage of classes before they can take the exam.

Write a function `can_take_exam(attendance, required_percentage=75)`.

**Input**
- `attendance`: a string with one character for each class held. `"P"` means the student was present and `"A"` means the student was absent. Only uppercase `P` and `A` are valid.
- `required_percentage`: a number from 0 to 100, inclusive, indicating the minimum attendance percentage. The default is 75.

**Output**
- Return `True` if the student's attendance percentage is greater than or equal to the required percentage; otherwise, return `False`.
- If `attendance` is empty, contains an invalid character, or `required_percentage` is outside the allowed range, print `Invalid input values.` and return `None`.

Assume the input types are correct. Each class has the same weight, and the admission decision must use the unrounded attendance percentage. Do not use lists, tuples, sets, or dictionaries.

```python
def can_take_exam(attendance, required_percentage=75):
    if attendance == "" or not 0 <= required_percentage <= 100:
        print("Invalid input values.")
        return

    present = 0
    for status in attendance:
        if status == "P":
            present += 1
        elif status != "A":
            print("Invalid input values.")
            return

    attendance_percentage = present / len(attendance) * 100
    return attendance_percentage >= required_percentage
```

```python
print(can_take_exam("PPAP"))      # Expected: True (75% attendance)
print(can_take_exam("PPAP", 80))  # Expected: False
print(can_take_exam("PAAA"))      # Expected: False (25% attendance)
print(can_take_exam("PPPP", 100)) # Expected: True

can_take_exam("")         # Expected: Invalid input values.
can_take_exam("PPXP")     # Expected: Invalid input values.
can_take_exam("PPAP", 110) # Expected: Invalid input values.
```

## Exercise 4: Plan your exam preparation

You are preparing for an exam and want to divide the reading over several days.

Write a function `study_plan(total_pages, pages_per_day)`.

**Input**
- `total_pages`: a positive integer indicating the number of pages to study.
- `pages_per_day`: a positive integer indicating the maximum number of pages you can study each day.

**Output**
- Print one line for each study day, starting from day 1. Each line must report the day number, the pages studied that day, and the pages still to study, using the format shown in the examples.
- Study `pages_per_day` pages each day, except on the last day if fewer pages remain. All pages must be covered, with no extra days after the reading is complete.
- Return the total number of days needed as an integer.
- If either input is less than or equal to 0, print `Invalid input values.` and return `None`, without printing a daily plan.

Assume the input types are correct. Do not use lists, tuples, sets, or dictionaries.

```python
def study_plan(total_pages, pages_per_day):
    if total_pages <= 0 or pages_per_day <= 0:
        print("Invalid input values.")
        return
    remaining = total_pages
    day = 0
    while remaining > 0:
        day += 1
        if remaining >= pages_per_day:
            today = pages_per_day
        else:
            today = remaining
        remaining -= today
        print(f"Day {day}: read {today} pages; {remaining} pages remaining.")
    return day
```

```python
days = study_plan(25, 10)
print(f"Days needed: {days}")
# Expected:
# Day 1: read 10 pages; 15 pages remaining.
# Day 2: read 10 pages; 5 pages remaining.
# Day 3: read 5 pages; 0 pages remaining.
# Days needed: 3

study_plan(8, 10)
# Expected: Day 1: read 8 pages; 0 pages remaining.

study_plan(0, 10)
# Expected: Invalid input values.
```

## Exercise 5: Running distance

Create a function `running_distance` that computes the distance covered by a runner who runs for a given duration at a given speed.

In input, the function takes:
- duration: the number of minutes run; it must be greater than 0
- pace_min_per_km: the speed, expressed as the number of minutes taken to cover 1km; it must be greater than 0
- pace_sec_per_km (optional): the number of seconds (to be added to pace_min_per_km); if given, it must be between 0 and 59 (defaults to 0)

If the input is valid, the function prints something like: "If you run at 4:30 min/km (i.e., 13.33 km/h) for 240 minutes, you will cover 53.33 kilometers."; otherwise, an error must be returned.

- The formula to calculate the distance in km is "duration in seconds" / "total number of seconds per km".
- The formula to calculate the speed in km/h is 3600 / "total number of seconds per km".
- Use the `round()` formula to round results to 2 decimals.

```python
def running_distance(duration, pace_min_per_km, pace_sec_per_km=0):
    if duration <= 0 or pace_min_per_km <= 0 or not 0 <= pace_sec_per_km <= 59:
        print("Invalid input values.")
        return

    pace_total_seconds = pace_min_per_km * 60 + pace_sec_per_km
    duration_seconds = duration * 60
    speed = 3600 / pace_total_seconds
    distance = duration_seconds / pace_total_seconds
    print(f"If you run at {pace_min_per_km}:{pace_sec_per_km:02d} min/km "
          f"(i.e., {round(speed, 2)} km/h) for {duration} minutes, "
          f"you will cover {round(distance, 2)} kilometers.")
```

```python
running_distance(240, 4, 30)
```

## Exercise 6: Running time

Create a function `running_time` that computes the pace you must have to complete a given distance in a given time.

In input, the function takes:
- distance: the expected distance in km; it must be greater than 0
- time: the expected time in minutes; it must be greater than 0

If the input is valid, the function prints something like: "If you want to run 42.195 km in 180 minutes, you should run at 14.06 km/h (with a pace of 4:15 min/km)."; otherwise, an error must be returned.

- The formula to calculate the speed in km/h is "distance in km" / "duration in hours".
- The formula to calculate the pace in min/km is "duration in seconds" / "distance in km". This gives you the result in sec/km; how to convert it to min/km?
- Use the `round()` formula to round results to 2 decimals.

```python
def running_time(distance, time):
    if distance <= 0 or time <= 0:
        print("Invalid input values.")
        return

    speed = distance / (time / 60)
    pace_total_seconds = time * 60 / distance
    # Keep full seconds, as in the original exercise's example.
    pace_minutes = int(pace_total_seconds // 60)
    pace_seconds = int(pace_total_seconds % 60)
    print(f"If you want to run {distance} km in {time} minutes, "
          f"you should run at {round(speed, 2)} km/h "
          f"(with a pace of {pace_minutes}:{pace_seconds:02d} min/km).")
```

```python
running_time(42.195, 180)
```

## Exercise 7: Running updates

Create a function that prints updates on the status of a race of a runner.

In input, the function takes:
- the speed of the runner in km/h
- the length of the race in km
- the frequency of updates in km

Depending on the frequency, the function must print updates on the distance covered by the runner.

```python
def race_updates(speed, length, frequency):
    if speed <= 0 or length <= 0 or frequency <= 0:
        print("Invalid input values.")
        return

    checkpoint = 1
    distance = checkpoint * frequency
    while distance < length:
        total_seconds = round(distance / speed * 3600)
        minutes = total_seconds // 60
        seconds = total_seconds % 60
        print(f"It took {minutes} min {seconds} sec for the runner to cover {distance} km")
        checkpoint += 1
        distance = checkpoint * frequency

    total_seconds = round(length / speed * 3600)
    minutes = total_seconds // 60
    seconds = total_seconds % 60
    print(f"Final time to complete {length} km: {minutes} min {seconds} sec")
```

```python
# Example usage:
race_updates(10,42,5)

# Expect output:
# It took 30 min 0 sec for the runner to cover 5 km
# It took 60 min 0 sec for the runner to cover 10 km
# It took 90 min 0 sec for the runner to cover 15 km
# It took 120 min 0 sec for the runner to cover 20 km
# It took 150 min 0 sec for the runner to cover 25 km
# It took 180 min 0 sec for the runner to cover 30 km
# It took 210 min 0 sec for the runner to cover 35 km
# It took 240 min 0 sec for the runner to cover 40 km
# Final time to complete 42 km: 252 min 0 sec
```
